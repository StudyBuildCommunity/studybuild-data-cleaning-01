```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("D:\projects\project-01-data-cleaning\data\cleaned_dataset.xlsx")
```

```python
# age distribution
plt.figure(figsize=(8,5))
plt.hist(df["age"],bins=10)
plt.title("age distribution")
plt.xlabel("age")
plt.ylabel("number of customers")
plt.show()
```

```python
# membership tier distribution
df["membership_tier"].value_counts().plot(kind="bar")

plt.title("Membership Tier Distribution")
plt.xlabel("Membership Tier")
plt.ylabel("Count")
plt.show()
```

```python
# payment methods

df["payment_method"].value_counts().plot(kind="bar")

plt.title("Payment Methods")
plt.xlabel("Payment Method")
plt.ylabel("Count")
plt.show()
```

```python
# device usage

df["device"].value_counts().plot(kind="bar")

plt.title("Device Usage")
plt.xlabel("Device")
plt.ylabel("Count")
plt.show()
```

```python
# Average Spending by Membership

df.groupby("membership_tier")["total_spending"].mean().plot(kind="bar")

plt.title("Average Spending by Membership")
plt.xlabel("Membership Tier")
plt.ylabel("Average Spending")
plt.show()
```
