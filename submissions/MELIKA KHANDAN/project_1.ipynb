```python
age_median = df["age"].median()

print("Median age:", age_median)
```

```python
# اصلاح سن غیرواقعی
df.loc[df["age"] > 100, "age"] = age_median

# پر کردن مقدار خالی age
df["age"] = df["age"].fillna(df["age"].median())
```

```python
invalid_satisfaction = df[
    ~df["satisfaction_score"].between(1, 5)
]

invalid_satisfaction
```

```python
categorical_columns = [
    "gender",
    "city",
    "province",
    "membership_tier",
    "payment_method",
    "device",
    "discount_used"
]

for col in categorical_columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
df["calculated_spending"] = (
    df["purchase_count"] * df["avg_order_value"]
)
```

```python
df["spending_difference"] = (
    df["total_spending"] - df["calculated_spending"]
)
```

```python
df[
    df["spending_difference"].abs() > 0.01
][[
    "customer_id",
    "purchase_count",
    "avg_order_value",
    "total_spending",
    "calculated_spending",
    "spending_difference"
]]
```

```python
# اصلاح total_spending اشتباه برای customer_id = 1030

df.loc[
    df["customer_id"] == 1030,
    "total_spending"
] = df.loc[
    df["customer_id"] == 1030,
    "calculated_spending"
].iloc[0]
```

```python
df[df["total_spending"].isna()]
```

```python
df.loc[
    df["total_spending"].isna(),
    "total_spending"
] = (
    df["purchase_count"] *
    df["avg_order_value"]
)
```

```python
df[df["customer_id"].isin([1030, 1040])][[
    "customer_id",
    "purchase_count",
    "avg_order_value",
    "total_spending",
    "calculated_spending"
]]
```

```python
df = df.drop(
    columns=["calculated_spending", "spending_difference"]
)
```

```python
print("Final shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
print("\nInvalid ages:")
print(df[~df["age"].between(18, 100)])

print("\nInvalid satisfaction scores:")
print(
    df[~df["satisfaction_score"].between(1, 5)]
)
```

```python
df.to_excel(
    "cleaned_dataset.xlsx",
    index=False
)
```
