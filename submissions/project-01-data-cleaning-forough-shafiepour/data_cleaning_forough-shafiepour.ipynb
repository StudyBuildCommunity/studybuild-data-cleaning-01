```python
# بررسی ناسازگاری قاعده total = count × avg
df['expected_total'] = df['purchase_count'] * df['avg_order_value']
#Calculate expected total based on the rule

df['diff'] = (df['total_spending'] - df['expected_total']).abs()
#Calculate absolute difference

df['rule_holds'] = df['diff'] < 1
#Check if the rule holds within 1 unit tolerance

print(f"Rows where rule holds: {df['rule_holds'].sum()} / {len(df)}")
#Count how many rows follow the rule

print(f"\nRows where rule is broken:")
print(df[~df['rule_holds']][['customer_id', 'purchase_count', 'avg_order_value', 
                              'total_spending', 'expected_total', 'diff']])
#Show rows where rule is violated
```

```python
print(f"total_spending stats:\n{df['total_spending'].describe()}")
#Statistical summary of total_spending

print(f"\nRow 1030 details:")
print(df[df['customer_id'] == 1030].T)
#Show row 1030 transposed for readability

print(f"\nTop 5 highest total_spending:")
print(df.nlargest(5, 'total_spending')[['customer_id', 'purchase_count', 
                                        'avg_order_value', 'total_spending',
                                        'membership_tier']])
#Show top 5 rows by total_spending
```

```python
# ─── IQR Method for Outlier Detection ───
Q1 = df['total_spending'].quantile(0.25)
#Calculate first quartile (25%)

Q3 = df['total_spending'].quantile(0.75)
#Calculate third quartile (75%)

IQR = Q3 - Q1
#Calculate interquartile range

lower_bound = Q1 - 1.5 * IQR
#Calculate lower bound

upper_bound = Q3 + 1.5 * IQR
#Calculate upper bound

print(f"Q1: {Q1:.2f}")
print(f"Q3: {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Lower Bound: {lower_bound:.2f}")
print(f"Upper Bound: {upper_bound:.2f}")
#Display all IQR metrics

outliers = df[(df['total_spending'] < lower_bound) | 
              (df['total_spending'] > upper_bound)].copy()
#Filter rows outside the bounds

outliers['expected_total'] = outliers['purchase_count'] * outliers['avg_order_value']
#Add expected_total column based on business rule

outliers['diff'] = (outliers['total_spending'] - outliers['expected_total']).abs()
#Calculate absolute difference from expected value

outliers['rule_holds'] = outliers['diff'] < 1
#Flag whether business rule holds (within 1 unit tolerance)

print(f"\nNumber of outliers: {len(outliers)}")
#Show count of outliers

# ─── Show outliers with RED highlighted total_spending ───
outliers[['customer_id', 'purchase_count', 'avg_order_value', 'expected_total',
          'total_spending', 'diff', 'rule_holds', 'membership_tier']].style \
    .map(lambda v: 'color: red; font-weight: bold', 
         subset=['total_spending']) \
    .format({'expected_total': '{:,.2f}', 
             'avg_order_value': '{:,.2f}',
             'total_spending': '{:,.2f}',
             'diff': '{:,.2f}'})
#Display outliers with total_spending in red and numbers formatted
```

```python
# ─── Fix row 1030: invalid outlier (business rule violation) ───
df.loc[df['customer_id'] == 1030, 'total_spending'] = 4079.14
#Correct invalid value to expected (26 × 156.89)

# ─── Fix row 1040: impute NaN using business rule ───
df.loc[df['customer_id'] == 1040, 'total_spending'] = (
    df.loc[df['customer_id'] == 1040, 'purchase_count'].values[0] *
    df.loc[df['customer_id'] == 1040, 'avg_order_value'].values[0]
)
#Impute missing value with count × avg_order_value

# ─── Verification ───
print(f"Remaining NaN in total_spending: {df['total_spending'].isnull().sum()}")
#Verify no NaN remain

print(f"New Max: {df['total_spending'].max():.2f}")
#Verify 25000 is gone

print(f"New Mean: {df['total_spending'].mean():.2f}")
#Show new mean after fix

print(f"\nRow 1030 after fix:")
print(df[df['customer_id'] == 1030][['customer_id', 'purchase_count', 
                                      'avg_order_value', 'total_spending']])
#Verify row 1030 is corrected

print(f"\nRow 1040 after fix:")
print(df[df['customer_id'] == 1040][['customer_id', 'purchase_count', 
                                      'avg_order_value', 'total_spending']])
#Verify row 1040 is imputed
```

```python
# ─── Recalculate IQR after fixing ───
Q1_new = df['total_spending'].quantile(0.25)
Q3_new = df['total_spending'].quantile(0.75)
IQR_new = Q3_new - Q1_new

lower_new = Q1_new - 1.5 * IQR_new
upper_new = Q3_new + 1.5 * IQR_new

print(f"New Q1: {Q1_new:.2f}")
print(f"New Q3: {Q3_new:.2f}")
print(f"New IQR: {IQR_new:.2f}")
print(f"New Lower Bound: {lower_new:.2f}")
print(f"New Upper Bound: {upper_new:.2f}")

new_outliers = df[(df['total_spending'] < lower_new) | 
                  (df['total_spending'] > upper_new)]
print(f"\nRemaining outliers: {len(new_outliers)}")
print(new_outliers[['customer_id', 'total_spending', 'membership_tier']])
#Check remaining outliers after fixing
```

```python
# ─── Verify row 1059 follows the business rule ───
print("Row 1059 details:")
print(df[df['customer_id'] == 1059][['customer_id', 'purchase_count', 
                                      'avg_order_value', 'total_spending']])
#Inspect row 1059

expected_1059 = df.loc[df['customer_id'] == 1059, 'purchase_count'].values[0] * \
                df.loc[df['customer_id'] == 1059, 'avg_order_value'].values[0]
print(f"\nExpected for 1059: {expected_1059:.2f}")
print(f"Actual for 1059: {df.loc[df['customer_id'] == 1059, 'total_spending'].values[0]:.2f}")
print(f"Rule holds: {abs(expected_1059 - df.loc[df['customer_id'] == 1059, 'total_spending'].values[0]) < 1}")
#Verify business rule for row 1059
```

```python
# ─── Step 2a: Drop helper columns (if exist) ───
helper_cols = ['expected_total', 'diff', 'rule_holds']
#List helper columns to remove

cols_to_drop = [c for c in helper_cols if c in df.columns]
#Filter only those that actually exist

if cols_to_drop:
    df = df.drop(columns=cols_to_drop)
    print(f"Dropped: {cols_to_drop}")
else:
    print("No helper columns to drop")
#Drop only existing columns

print(f"\nShape after dropping: {df.shape}")
#Verify dimensions

print(f"Columns now ({len(df.columns)}): {list(df.columns)}")
#List remaining columns
```

```python
# ─── Step 2b: Convert signup_date to datetime ───
df['signup_date'] = pd.to_datetime(df['signup_date'], errors='coerce')
#Convert string dates to datetime objects

print(f"signup_date dtype: {df['signup_date'].dtype}")
#Verify conversion

print(f"signup_date range: {df['signup_date'].min()} to {df['signup_date'].max()}")
#Check date range

print(f"Any failed conversions (NaT): {df['signup_date'].isnull().sum()}")
#Check if any dates failed to parse

print(f"\nFirst 5 rows of signup_date:")
print(df['signup_date'].head())
#Preview converted dates
```

```python
# ─── Step 2c: Convert age to nullable Int64 ───
print(f"Before: dtype = {df['age'].dtype}")
#Check current dtype

print(f"Before: sample = {df['age'].head(3).tolist()}")
#Show sample values (note the .0)

df['age'] = df['age'].astype('Int64')
#Convert to nullable integer

print(f"\nAfter: dtype = {df['age'].dtype}")
#Verify new dtype

print(f"After: sample = {df['age'].head(3).tolist()}")
#Show sample values (no .0)

print(f"\nMissing values in age: {df['age'].isnull().sum()}")
#Verify NaN is preserved
```

```python
# ─── Step 2d: Convert discount_used to boolean ───
print(f"Before: {df['discount_used'].unique()}")
#Show unique values before conversion

df['discount_used'] = df['discount_used'].map({'Yes': True, 'No': False})
#Map Yes/No to True/False

print(f"After: {df['discount_used'].unique()}")
#Show unique values after conversion

print(f"dtype: {df['discount_used'].dtype}")
#Verify boolean dtype

print(f"\nCount of True: {df['discount_used'].sum()}")
#Count how many customers used discount
```

```python
# ─── Step 2e: Convert nominal columns to category ───
nominal_cols = ['gender', 'city', 'province', 'payment_method', 'device']
#List of nominal columns

print("Before:")
for col in nominal_cols:
    print(f"  {col}: {df[col].dtype} | unique: {df[col].nunique()}")
#Show dtypes and unique counts before

for col in nominal_cols:
    df[col] = df[col].astype('category')
#Convert each nominal column to category

print("\nAfter:")
for col in nominal_cols:
    print(f"  {col}: {df[col].dtype} | unique: {df[col].nunique()}")
#Show dtypes after conversion

print(f"\nMemory usage: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
#Check total memory usage
```

```python
# ─── Step 2f: Convert membership_tier to ordered category ───
from pandas.api.types import CategoricalDtype
#Import CategoricalDtype for custom ordering

print(f"Before: {df['membership_tier'].dtype}")
print(f"Values: {df['membership_tier'].unique()}")
#Show current state

tier_order = CategoricalDtype(
    categories=['Bronze', 'Silver', 'Gold', 'VIP'],
    ordered=True
)
#Define the correct hierarchy

df['membership_tier'] = df['membership_tier'].astype(tier_order)
#Convert to ordered category

print(f"\nAfter: {df['membership_tier'].dtype}")
print(f"Order: {df['membership_tier'].cat.categories.tolist()}")
#Show the defined order

print(f"\nTest sorting by tier:")
print(df.sort_values('membership_tier')['membership_tier'].head(10).tolist())
#Verify sorting follows hierarchy
```

```python
# ─── Final dtype verification ───
print("Final data types:")
print(df.dtypes)
#Show all column dtypes

print(f"\nDetailed info:")
df.info()
#Show detailed info

print(f"\nShape: {df.shape}")
#Verify dimensions

print(f"\nMissing values:")
print(df.isnull().sum())
#Verify missing values

print(f"\nDuplicates: {df.duplicated(subset='customer_id').sum()}")
#Verify no duplicates
```
