### تغییر تایپ ویژگی ها به تایپ استاندارد به صورت امن

```python

df['gender'] = df['gender'].astype('string')

df['age'] = (pd.to_numeric(df['age'],errors='coerce').astype('Int64'))

df['city'] = df['city'].astype('string')

df['signup_date'] = pd.to_datetime(df['signup_date'], format='%Y-%m-%d', errors='coerce')

tier_dtype = pd.CategoricalDtype(categories=['Bronze', 'Silver', 'Gold', 'VIP'], ordered=True)
df['membership_tier'] = df['membership_tier'].astype(tier_dtype) 

df['payment_method'] = df['payment_method'].astype('string')

df['device'] = df['device'].astype('string')

df['discount_used'] = df['discount_used'].astype('string')
```

# Checking and Filling Missing Values

### age

```python
print('min :', df['age'].min())
print('max :' ,df['age'].max())
print('average :', df['age'].mean())
```

```python
# age

df[df['age'] >100]
```

```python
# 1) پیدا کردن ردیف‌های مشکوک
invalid_age_index = df.index[df['age'] > 100]

# 2) ذخیره رکوردها قبل از تغییر
invalid_age_rows_before = df.loc[invalid_age_index].copy()

# 3) تبدیل سن نامعتبر به مقدار گمشده
df.loc[invalid_age_index, 'age'] = np.nan

# 4) دیدن رکوردها بعد از تغییر
invalid_age_rows_after = df.loc[invalid_age_index].copy()

display(invalid_age_rows_before)
display(invalid_age_rows_after)
```

```python
# fill by mean

df['age'] = df['age'].fillna(df['age'].mean().round())
df['age'].isna().sum()
```

### total_spending

```python
print('min :', df['total_spending'].min())
print('max :' ,df['total_spending'].max())
print('average :', df['total_spending'].mean())
```

```python
df['exp_spending'] = df['purchase_count'] * df['avg_order_value']
df['diff'] = df['total_spending'] - df['exp_spending']
df[df['diff'] > 0.01]
```

```python
# 1) پیدا کردن ردیف‌های مشکوک
invalid_total_index = df.index[df['total_spending'] == 25000]

# 2) ذخیره رکوردها قبل از تغییر
invalid_total_rows_before = df.loc[invalid_total_index].copy()

# 3) تبدیل سن نامعتبر به مقدار گمشده
df.loc[invalid_total_index, 'total_spending'] = np.nan

# 4) دیدن رکوردها بعد از تغییر
invalid_total_rows_after = df.loc[invalid_total_index].copy()

display(invalid_total_rows_before)
display(invalid_total_rows_after)
```
